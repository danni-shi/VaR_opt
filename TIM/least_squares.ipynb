```python
import numpy as np
from scipy.optimize import least_squares
import arviz as az
import pickle
from tqdm import tqdm
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_results = pd.read_csv('/nfs/home/dannis/AC-model/TIM/TIM_simulated_data/20250826/least_squares_results/summary.csv')
```

```python

fig, axes = plt.subplots(3, 1, figsize=(8, 14), sharex=True)
true_params = {'gamma': 1e-4, 'kappa': 1e-4, 'rho': 2.231}
for ax, param in zip(axes, ['gamma_hat', 'kappa_hat', 'rho_hat']):
    ax.set_title(param)
    sns.lineplot(data=df_results, x='n_trades', y=param, ax=ax)
    ax.axhline(true_params[param.split('_')[0]], color='red', linestyle='--', label='true value')
    ax.ticklabel_format(style='scientific', axis='y', scilimits=(0,0))
    ax.grid()

fig.suptitle('Convergence of Least Squares Estimates', fontsize=16)

plt.savefig('least_squares_estimates.png')
plt.show()
```
